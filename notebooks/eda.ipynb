```python
#Channel Analysis
```

```python
#Top 10 Channels by Views
```

```python
top_channels = (
    df.groupby("channelName")["viewCount"]
      .sum()
      .sort_values(ascending=False)
      .head(10)
)

plt.figure(figsize=(10,6))
plt.bar(top_channels.index, top_channels.values)
plt.xticks(rotation=45, ha="right")
plt.title("Top 10 Channels by Total Views")
plt.ylabel("Views")
plt.show()
```

```python
#Top 10 Channels by Average Engagement
```

```python
engagement = (
    df.groupby("channelName")["engagement_rate"]
      .mean()
      .sort_values(ascending=False)
      .head(10)
)

plt.figure(figsize=(10,6))
plt.bar(engagement.index, engagement.values)
plt.xticks(rotation=45, ha="right")
plt.title("Top Channels by Average Engagement")
plt.show()
```

```python
#Time-Based Analysis
```

```python
# Upload Trend by Year
```

```python
yearly = df.groupby("upload_year").size()

plt.figure(figsize=(8,5))
plt.plot(yearly.index, yearly.values, marker="o")
plt.title("Videos Uploaded by Year")
plt.xlabel("Year")
plt.ylabel("Videos")
plt.show()
```

```python
#Upload Trend by Month
```

```python
monthly = df.groupby("upload_month").size()

plt.figure(figsize=(8,5))
plt.plot(monthly.index, monthly.values, marker="o")
plt.title("Videos Uploaded by Month")
plt.xlabel("Month")
plt.ylabel("Videos")
plt.show()
```

```python
#Upload Trend by Hour
```

```python
hourly = df.groupby("upload_hour").size()

plt.figure(figsize=(8,5))
plt.bar(hourly.index, hourly.values)
plt.title("Upload Distribution by Hour")
plt.xlabel("Hour")
plt.ylabel("Videos")
plt.show()
```
