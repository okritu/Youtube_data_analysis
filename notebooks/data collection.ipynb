```python
df_clean = df.copy()
```

```python
df_clean.dtypes
```

```python
numeric_cols = ['viewCount', 'likes', 'commentsCount']

for col in numeric_cols:
    df_clean[col] = pd.to_numeric(df_clean[col], errors='coerce')
```

```python
df_clean.dtypes
```

```python
df_clean['date'] = pd.to_datetime(df_clean['date'], errors='coerce')
```

```python
df_clean['date'].head()
```

```python
df_clean.isnull().sum()
```

```python
df_clean['url'].duplicated().sum()
```

```python
df_clean['upload_year'] = df_clean['date'].dt.year
```

```python
df_clean['upload_month'] = df_clean['date'].dt.month_name()
```

```python
df_clean['upload_day'] = df_clean['date'].dt.day_name()
```

```python
df_clean['upload_hour'] = df_clean['date'].dt.hour
```

```python
df_clean['engagement_rate'] = (
    (df_clean['likes'] + df_clean['commentsCount'])
    / df_clean['viewCount']
) * 100
```

```python
df_clean['like_ratio'] = (
    df_clean['likes']
    / df_clean['viewCount']
) * 100
```

```python
df_clean['comment_ratio'] = (
    df_clean['commentsCount']
    / df_clean['viewCount']
) * 100
```

```python
df_clean['popularity_score'] = (
      df_clean['viewCount'] * 0.5
    + df_clean['likes'] * 0.3
    + df_clean['commentsCount'] * 0.2
)
```
